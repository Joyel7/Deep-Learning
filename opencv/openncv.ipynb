```python
import cv2
import matplotlib.pyplot as plt
```

```python
img=cv2.imread(r'C:\Users\joyel\Downloads\og2.jpg')
conv_imgg3=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(conv_imgg3)
```

```python
cv2.rectangle(conv_imgg3,pt1=(250,5),pt2=(600,370),color=(255,0,0),thickness=3)
plt.imshow(conv_imgg3)
```
