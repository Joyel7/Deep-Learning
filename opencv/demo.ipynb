```python
#greyscale conversion
imagegrey=cv2.imread(r'C:\Users\joyel\Downloads\ba1.webp',cv2.IMREAD_GRAYSCALE)
```

```python
#show
cv2.imshow('BABY',imagegrey)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
#convert to binary
thresh,imagebinary=cv2.threshold(imagegrey,125,255,cv2.THRESH_BINARY)
```

```python
#show
cv2.imshow('BABY',imagebinary)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
#convert to binary#machine itself
thresh,imagebinary=cv2.threshold(imagegrey,125,255,cv2.THRESH_OTSU)
```

```python
#show
cv2.imshow('BABY',imagebinary)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
cv2.imwrite('imgggg1.jpeg',imagebinary)
```
