```python
import cv2
```

```python
eye_data_model=cv2.CascadeClassifier(r'D:\vs_code_dl\eyedetection\haarcascade_eye.xml')
eye_data_model
```

```python
#video
video=cv2.VideoCapture(r'D:\vs_code_dl\eyedetection\WhatsApp Video 2025-11-26 at 12.52.38 PM.mp4') 
                       #0 for own machine web cam .1 for external webcam
#to read video
while True:
    success,img=video.read()
    if success:
        gray_img=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
        faces=eye_data_model.detectMultiScale(gray_img,minNeighbors=35)
        for(x,y,w,h) in faces:
            cv2.rectangle(img,(x,y),(x+w,y+h),(0,0,255),4)




        cv2.imshow('video',img)
        if cv2.waitKey(1) & 0xFF==ord('q'):
            break
    else:
     print("Finished!!!")
     break
video.release()
cv2.destroyAllWindows
```
