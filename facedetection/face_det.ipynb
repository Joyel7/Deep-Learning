```python
import cv2
import matplotlib.pyplot as plt
```

```python
img1=cv2.imread(r'D:\vs_code_dl\facedetection\cristiano-ronaldo-rooney-manchester-united-old-trafford-wallpaper-preview.jpg')
img1
```

```python
con_img1=cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
plt.imshow(con_img1)
```

```python
face_data=cv2.CascadeClassifier(r'D:\vs_code_dl\facedetection\Unconfirmed 250836.crdownload')
face_data
```

```python
faces=face_data.detectMultiScale(con_img1,minNeighbors=5)
faces
```

```python
for [x,y,w,h] in faces:
    cv2.rectangle(con_img1,pt1=(x,y),pt2=(x+w,y+h),color=(255,0,0),thickness=3)
```

```python
plt.imshow(con_img1)
```
